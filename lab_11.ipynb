```python
from IPython.display import HTML, display

html_code = r"""
<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<title>ColorWave Accessibility Studio</title>

<style>
:root {
  --bg: #f4f0ff;
  --card: #ffffff;
  --text: #242044;
  --accent: #7048e8;
  --focus: #ff7b00;
}

* { box-sizing: border-box; }

body {
  margin: 0;
  padding: 25px;
  font-family: Arial, sans-serif;
  background: var(--bg);
  color: var(--text);
  transition: background .25s;
}

header {
  background: linear-gradient(135deg, #7048e8, #e64980);
  color: white;
  padding: 28px;
  border-radius: 18px;
  margin-bottom: 22px;
}

h1 { margin-top: 0; }
h2 { color: var(--accent); }

.card {
  background: var(--card);
  padding: 22px;
  border-radius: 15px;
  margin: 18px 0;
  box-shadow: 0 5px 18px #30206018;
}

button, input, select {
  font: inherit;
  padding: 11px 14px;
  margin: 5px;
  border-radius: 8px;
}

button {
  background: var(--accent);
  color: white;
  border: 2px solid var(--accent);
  cursor: pointer;
}

button:hover { opacity: .85; }

button:focus-visible,
input:focus-visible,
select:focus-visible,
[tabindex="0"]:focus-visible {
  outline: 4px solid var(--focus);
  outline-offset: 4px;
}

.tabs {
  display: flex;
  flex-wrap: wrap;
  gap: 8px;
}

.tabs button[aria-selected="true"] {
  background: #252044;
  border-color: #252044;
}

.panel[hidden] { display: none; }

label { font-weight: bold; }

input[type="color"] {
  width: 60px;
  height: 45px;
  padding: 3px;
  vertical-align: middle;
}

.notice {
  padding: 12px;
  border-left: 5px solid var(--accent);
  background: #eee8ff;
  color: #242044;
  margin-top: 12px;
  border-radius: 6px;
}

dialog {
  border: 3px solid var(--accent);
  border-radius: 16px;
  padding: 25px;
  max-width: 420px;
}

dialog::backdrop { background: #15102a99; }

@media (max-width: 600px) {
  body { padding: 12px; }
  header { padding: 20px; }
}
</style>
</head>

<body>
<header>
  <h1>🎨 ColorWave Accessibility Studio</h1>
  <p>Explore accessible design, keyboard navigation,
  custom colors, focus indicators and interactive controls.</p>
</header>

<section class="card">
  <h2>Customize Your Website</h2>

  <label for="bgColor">Background color:</label>
  <input type="color" id="bgColor" value="#f4f0ff">

  <label for="accentColor">Button color:</label>
  <input type="color" id="accentColor" value="#7048e8">

  <label for="focusColor">Focus indicator:</label>
  <input type="color" id="focusColor" value="#ff7b00">

  <button id="resetColors">Reset Colors</button>
  <p id="colorStatus" role="status" aria-live="polite"></p>
</section>

<section class="card">
  <h2>Keyboard Navigation</h2>
  <p>Focus a module button, then use Left and Right Arrow keys.
  Home selects the first module and End selects the last.</p>

  <div class="tabs" role="tablist" aria-label="Studio modules">
    <button role="tab" id="tab1" aria-controls="panel1"
      aria-selected="true" tabindex="0">1. Focus Lab</button>
    <button role="tab" id="tab2" aria-controls="panel2"
      aria-selected="false" tabindex="-1">2. Color Lab</button>
    <button role="tab" id="tab3" aria-controls="panel3"
      aria-selected="false" tabindex="-1">3. Audit Lab</button>
  </div>

  <div id="panel1" class="panel" role="tabpanel"
    aria-labelledby="tab1" tabindex="0">
    <h2>Focus Indicator Lab</h2>
    <p>Use Tab to focus the buttons. A bright outline shows
    which control is currently focused.</p>
    <button class="popupButton">Open Information Popup</button>
    <button class="demoButton">Test Focus Button</button>
    <p id="focusStatus" role="status" aria-live="polite"></p>
  </div>

  <div id="panel2" class="panel" role="tabpanel"
    aria-labelledby="tab2" tabindex="0" hidden>
    <h2>Color Lab</h2>
    <p>Change the colors above and see the design update instantly.</p>
    <button class="demoButton">Color Test Button</button>
    <p>Remember: text must remain readable against its background.</p>
  </div>

  <div id="panel3" class="panel" role="tabpanel"
    aria-labelledby="tab3" tabindex="0" hidden>
    <h2>Accessibility Audit</h2>
    <label><input type="checkbox"> Visible focus indicator</label><br>
    <label><input type="checkbox"> Buttons have clear labels</label><br>
    <label><input type="checkbox"> Text is readable</label><br>
    <label><input type="checkbox"> Keyboard navigation works</label><br>
    <button id="auditButton">Show Audit Summary</button>
    <p id="auditStatus" role="status" aria-live="polite"></p>
  </div>
</section>

<section class="card">
  <h2>Interactive Controls</h2>
  <p>Click the buttons to test their actions and feedback.</p>
  <button id="helloButton">Click Me</button>
  <p id="message" class="notice" role="status" aria-live="polite">
    Your interaction message will appear here.
  </p>
</section>

<dialog id="infoDialog" aria-labelledby="dialogTitle">
  <h2 id="dialogTitle">✨ Accessibility Popup</h2>
  <p>This popup demonstrates an interactive dialog.
  Press Escape or select Close to return to the page.</p>
  <button id="closeDialog">Close Popup</button>
</dialog>

<script>
const root = document.documentElement;

const bgColor = document.getElementById("bgColor");
const accentColor = document.getElementById("accentColor");
const focusColor = document.getElementById("focusColor");

function applyColors() {
  root.style.setProperty("--bg", bgColor.value);
  root.style.setProperty("--accent", accentColor.value);
  root.style.setProperty("--focus", focusColor.value);
  document.getElementById("colorStatus").textContent =
    "Colors updated successfully.";
}

[bgColor, accentColor, focusColor].forEach(input => {
  input.addEventListener("input", applyColors);
});

document.getElementById("resetColors").addEventListener("click", () => {
  bgColor.value = "#f4f0ff";
  accentColor.value = "#7048e8";
  focusColor.value = "#ff7b00";
  applyColors();
});

const tabs = [...document.querySelectorAll('[role="tab"]')];

function selectTab(tab, moveFocus = false) {
  tabs.forEach(item => {
    const selected = item === tab;
    item.setAttribute("aria-selected", String(selected));
    item.tabIndex = selected ? 0 : -1;
    document.getElementById(
      item.getAttribute("aria-controls")
    ).hidden = !selected;
  });

  if (moveFocus) tab.focus();
}

tabs.forEach(tab => {
  tab.addEventListener("click", () => selectTab(tab));

  tab.addEventListener("keydown", event => {
    let index = tabs.indexOf(tab);

    if (event.key === "ArrowRight") index = (index + 1) % tabs.length;
    else if (event.key === "ArrowLeft")
      index = (index - 1 + tabs.length) % tabs.length;
    else if (event.key === "Home") index = 0;
    else if (event.key === "End") index = tabs.length - 1;
    else return;

    event.preventDefault();
    selectTab(tabs[index], true);
  });
});

const dialog = document.getElementById("infoDialog");
const popupButton = document.querySelector(".popupButton");

popupButton.addEventListener("click", () => dialog.showModal());

document.getElementById("closeDialog").addEventListener("click", () => {
  dialog.close();
  popupButton.focus();
});

document.querySelectorAll(".demoButton").forEach(button => {
  button.addEventListener("click", () => {
    document.getElementById("focusStatus").textContent =
      "You selected: " + button.textContent;
  });
});

document.getElementById("helloButton").addEventListener("click", () => {
  document.getElementById("message").textContent =
    "🎉 Button clicked! Your interaction worked successfully.";
});

document.getElementById("auditButton").addEventListener("click", () => {
  const checks = [...document.querySelectorAll(
    '#panel3 input[type="checkbox"]'
  )];

  const complete = checks.filter(check => check.checked).length;

  document.getElementById("auditStatus").textContent =
    "You checked " + complete + " of " + checks.length +
    " accessibility items.";
});
</script>
</body>
</html>
"""

display(HTML(html_code))
```
